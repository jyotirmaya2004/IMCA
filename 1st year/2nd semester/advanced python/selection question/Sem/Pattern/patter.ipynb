```python
n=5
for i in range(1,n+1):
	for k in range(i,n):
		print(" ",end=" ")
	for j in range(i,0,-1):
		print("%c"%(i+64),end=" ")
	for l in range(2,i+1):
		print("%c"%(i+64),end=" ")
	print("\n")
for i in range(n-1,0,-1):
	for k in range(0,n-i):
		print(" ",end=" ")
	for j in range(i,0,-1):
		print("%c"%(i+64),end=" ")
	for l in range(i,1,-1):
		print("%c"%(i+64),end=" ")
	print("\n")
```
