# series

```python
import numpy as np
import pandas as pd
```

### Series is a one-dimensional labeled array capable of holding any data type (integers, strings, floating point numbers, Python objects, etc.). The axis labels are collectively referred to as the index. The basic method to create a Series is to call:

### s = pd.Series(data, index=index)
### Here, data can be many different things:

###  1. a Python dict

###  2. an ndarray

###  3. a scalar value (like 5)

### The passed index is a list of axis labels. Thus, this separates into a few cases depending on what data is:

```python
s= pd.Series([3,5,4,7,8,5,4,7])
```

```python
print(s)
```

```python


s = pd.Series(np.random.randn(5), index=["a", "b", "c", "d", "e"])
```

```python
print(s)
```

```python
s.index
```

```python
pd.Series(np.random.randn(5))
```

### From dict

#### Series can be instantiated from dicts:

```python
d = {"b": 1, "a": 0, "c": 2}
```

```python
pd.Series(d)
```

```python
d = {"a": 0.0, "b": 1.0, "c": 2.0}
```

```python
pd.Series(d)
```

```python
pd.Series(d, index=["b", "c", "d", "a"])
```

#### NaN (not a number) is the standard missing data marker used in pandas.

## From scalar value

#### If data is a scalar value, an index must be provided. The value will be repeated to match the length of index.

```python
pd.Series(5.0, index=["a", "b", "c", "d", "e"])
```

```python
pd.Series("Jyotirmaya", index=["a", "b", "c", "d", "e"])
```

## Name attribute
#### Series also has a name attribute:

```python
s=pd.Series([1,2,4,5,3],name="Something")
```

```python
s
```

#### The Series name can be assigned automatically in many cases, in particular, when selecting a single column from a DataFrame, the name will be assigned the column label.
