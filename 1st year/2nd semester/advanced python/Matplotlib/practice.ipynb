### Histogram

```python
dataset=np.random.randn(1000)
print(dataset)
plt.hist(dataset,bins=100,color='blue',edgecolor='black')
plt.xlabel("value")
plt.ylabel("Frequency")
plt.title("Histograph")
plt.show()
```

```python
dataset={"Year":[2004,2005,2006,2007,2008,2009,2010],
		 "percentage":[87,89,67,56,98,90,91]}
df=pd.DataFrame(dataset)
df.plot.bar(x="Year",y="percentage",color="red",title="result analysis")
plt.show()
```

```python
dataset={"Year":[2004,2005,2006,2007,2008,2009,2010],
		 "percentage":[87,89,67,56,98,90,91],
		 "no of Student":[32,65,32,54,75,54,75]}
df=pd.DataFrame(dataset)
df.plot.bar(x="Year",y=["percentage","no of Student"],color=["red","yellow"],edgecolor="black",title="result analysis")
plt.show()
```
