# viewing and copying array

```python
a=np.arange(1,16)
```

```python
print(a)
```

```python
b=a.view
```

```python
print(b)
```

```python
b=a.view()
```

```python
print(b)
```

```python
b[0]=9999
```

```python
print(a)
print(b)
```

### Both changed original and viewded

```python
c=a.copy()
```

```python
print(c)
```

```python
c[0]=1
```

```python
print(a)
print(c)
```

# Dimension of array

## ndim attribute

```python
print(a.ndim)
```

```python
arr=np.array([[1,2,4,2,3],[3,4,2,6,7]])
print(arr.ndim)
```

## shape attribute

```python
print(arr.shape)
```

#### change of array shape

```python
arr.shape=(5,2)
```

```python
print(arr)
```

## size attribute

```python
print(arr.size)
```

## itemsize attribute

#### it gives the memory size of the array

```python
print(arr.itemsize)
```

## dtype attribute

```python
print(arr)
print(arr.dtype)
```

## reshape() method

```python
arr1=np.arange(1,11)
print(arr1)
```

```python
arr1=arr1.reshape(2,5)
```

```python
print(arr1)
```

```python
arr1=arr1.reshape(5,2)
```

```python
print(arr1)
```
